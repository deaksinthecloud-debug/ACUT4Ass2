```python
import pandas as pd

df = pd.read_csv("../data/synthetic_clinical_dataset.csv")

# nTaking 20 random patients (fixed seed, so it's the same 20 every time)
new = df.sample(20, random_state=1).copy()

# Remove the outcome columns - new patients at discharge don't have them yet
new = new.drop(columns=["readmission_30d", "mortality"])

# Making bad data
new.iloc[0, new.columns.get_loc("age")] = 130
new.to_csv("../data/new_patients_example.csv", index=False)

print(new.shape)
new.head()
```
